### What does the detector measure?

Using the group's definitions, $r_t$ is the vector of per-step log returns. For the cash entry, this is zero. A fixed portfolio direction $J$ it unit length and has weights that sum to zero (therefore we call it zero-cost). We define its projected return as $h_t = J^T r_t$. Additionally, let $\mu$ be the common mean return vector (do not confuse this with asset index $\mu$) and let $\Gamma _1 = \operatorname{Cov}(r_{t+1},r_t)$ (this is the lag-1 autocovariance matrix). The definition of expected value covariance gives:

$$
\begin{aligned}
D(J)&=\mathbb E[h_t h_{t+1}]\\
&=\mathbb E[h_t]\mathbb E[h_{t+1}]+\operatorname{Cov}(h_{t+1},h_t)\\
&=(J^T\mu)^2+J^T\Gamma_1J\\
&=J^T(\Gamma_1+\mu\mu^T)J.
\end{aligned}
$$

This is claim (C6) under the fixed-portfolio assumption. For $k$ fixed orthonormal portfolio columns in a matrix $J$, we sum over the colums to get $\operatorname{tr}[J^T(\Gamma_1+\mu\mu^T)J]$. This is because the diagonals of $\Gamma_1$ gives the own-lag terms, and the off-diagonal entries give the cross-asset terms. Because we are only using the symmetric part of $\Gamma_1$ we cannot identify which asset leads.

Vásquez and Farinelli's (Eq. 90), the formula for the metric of arbitrage, is of the form $a^2 + a(b-a)$ and therefore reduces to the product $ab$. This is why we multiply consecutive projected returns $h_t$. But why do we multiply adjacent returns rather than squaring a single return? Squaring one return adds variance to its mean, but multiplying two returns replaces that variance term with a lag-1 covariance.

Let's consider the following scenarios to digest these concepts: if we recover a mean of 1 with a zero lag covariance this gives $D = 1$. If we have a zero mean with lag covariance of 1, this also gives $D = 1$. A negative $D$ means the reversal of drift direction outweighs the squared mean. A $D$ of zero can mean the terms cancel. Theoretically this means the detector cannot distinguish between covariance versus positive drift. However this is not to say the empirical detector is without merit. Their formula returns a drift score as intended, it just also includes an additional autocovariance term which they do not discuss.
