# Day 2 (10/20)
Definition of cointegration: The non stationary series $X$, $Y$ are cointegrated if $\exists \beta$ such that $X-\beta Y$ is stationary (mean is fixed, variance does not change with time). Equivalently, $X$ and $Y$ are cointegrated iff $z=X-\beta Y$ is mean reverting. (here we confirm this with adfuller)

```python
rng = np.random.default_rng(seed=0)
```

```python
#notes for future reference
#this stuff should probably get moved to its own class cuz were gonna be using it a lot (also abm and gbm functions later)

#decorator for different time series
def make_series(f):
    sig = inspect.signature(f) #basically this is just so I can use this on any function with parameters T and n
    wants_t = "t" in sig.parameters #in the case the function requires t (abm and gbm)
    @wraps(f)
    def wrapper(*args, **kwargs):
        bound = sig.bind(*args, **kwargs)  
        bound.apply_defaults()              
        #get the T and n
        T = bound.arguments["T"]
        n = bound.arguments["n"]
        t = np.linspace(0, T, n + 1) #create the array of times (intervals of dt)
        if wants_t and "t" not in kwargs:
            kwargs["t"] = t
        return pd.DataFrame({'Value':f(*args, **kwargs), 'Time':t}) #return dataframe
    return wrapper

#Brownian motion W(t)
def W_t(T, n):
    dt = T/n
    dW = np.sqrt(dt) * rng.standard_normal(n) # increments ~ N(0, dt) (required for Brownian motion)
    W = np.concatenate([[0], np.cumsum(dW)]) #running total and concatenate to 0
    return W

#Random walk
@make_series
def random_walk(T,n):
    return W_t(T, n)

#create mean-reverting noise
def mr_noise(T,n,rho):
    dt = T/n 
    t = np.linspace(0, T, n + 1)
    e = np.sqrt(dt) * rng.standard_normal(n)
    #create mean-reverting noise
    a=np.array([0])
    for i in range(n):
        a=np.append(a, rho*a[i]+e[i]) #def of mean reversion
    return a
```

```python
#Parameters
T=100
n=100
b=2
rho=0.5
dt=T/n

#create initial random walk (X and Y are related by this)
w=random_walk(T, n)

#copy the random walk to two different variables
X=w.copy()
Y=w.copy()

#multiply values of Y by b
Y['Value']=Y['Value']*b

#add mean reverting noise
X['Value']=X['Value']+mr_noise(T,n,rho)
Y['Value']=Y['Value']+mr_noise(T,n,rho)

fig, ax=plt.subplots(2,1,figsize=(6, 10))
ax[0].plot(X['Time'], X['Value'])
ax[0].plot(Y['Time'], Y['Value'])
ax[0].set_xlabel("Time")
ax[0].set_ylabel("Price")
ax[0].set_title("Price vs. Time for two cointegrated series")

series=(X['Value'])-(1/b)*Y['Value']
ax[1].plot(X['Time'], series)
ax[1].set_xlabel("Time")
ax[1].set_ylabel("Price")
ax[1].set_title("X-(1/b)Y vs. time (mean reverting noise)")
```

Prediction: It should be stationary because we defined it so that the two processes are based on the same random walk with mean reverting noise, so $X-\frac{1}{b}Y$ should be stationary.

```python
#calculate adf p value
result = adfuller(series, autolag='AIC')
print(f"P value: {result[1]}")
```

By the definition of the ADF test, the null hypothesis is that the time series is non stationary. Since $p<0.05$, we reject the null hypothesis (I don't know how much this should actually be, but I'm going to leave it at 0.05 for now). There is convincing statistical evidence that the time series $z=X-\frac{1}{b}Y$ is stationary. By the definition of cointegration, $X$ and $Y$ are cointegrated.

We have
$$\sigma_i=\begin{pmatrix}
0\\ \sigma_1 \\ \sigma_2
\end{pmatrix}$$
Since we know
$$\Omega_{\mu\nu}=\sum_a \sigma_{\mu}^a \sigma_{\nu}^a$$
We have
$$\Omega=\begin{pmatrix}
0&0&0\\
0&\sigma_1^2 & \sigma_1\sigma_2\\
0&\sigma_1\sigma_2 & \sigma_2^2
\end{pmatrix}$$
From Equation (20), 
$$J=\frac{1}{\sqrt{2}\sqrt{\sigma_1^2+\sigma_2^2-\sigma_1\sigma_2}}\begin{pmatrix}
\sigma_1-\sigma_2\\
\sigma_2\\
-\sigma_1
\end{pmatrix}
$$
The factor in the front is a constant. Let it be equal to $c$. Then, $$\sum_{\mu}J_{\mu}=c(\sigma_1-\sigma_2+\sigma_2-\sigma_1)=0$$ We also need to show $J\cdot \sigma=0$. Substituting in values,
\begin{align*}
&J\cdot \sigma\\
&\Rightarrow c\begin{pmatrix}\sigma_1-\sigma_2\\ \sigma_2\\ -\sigma_1\end{pmatrix}\cdot \begin{pmatrix}0\\ \sigma_1 \\ \sigma_2\end{pmatrix}\\
&\Rightarrow c(0+\sigma_1\sigma_2 - \sigma_1\sigma_2)=0
\end{align*}

Let $L$ be the leader asset and $F$ be the follower asset. V&F longs $F$ and shorts $L$. If $\hat{\alpha}(t)$ is the spread's return at each step, then the estimator gives $$\hat{\mathcal{A}}^2(t+1)=\hat{\alpha}(t)\hat{\alpha}(t+1)$$
I'm going to implement the calculation of the $J^A$ vectors so we can compare our results from the simulation with V&F figure 1 (20 log normal random walks)

```python
#arithmetic brownian motion
@make_series
def abm(T, n, alpha, sigma, X_0, t=None):
    return X_0+(alpha*t)+(sigma*W_t(T, n))

#geometric brownian motion (log normal random walks)
@make_series
def gbm(T, n, alpha, sigma, X_0, t=None):
    return X_0*np.exp((alpha-(0.5*sigma**2))*t+(sigma*W_t(T,n)))

#v&f equation 2 SDE solution (equation 106)
#alpha_mu is a 1d array, sigma (sigma_mu^a) is 2d array (mu: 1 to N=row, a: 1 to d=column)
#pls tell me if theres a way to make this more efficient :skull:
@make_series
def X_mu(T, n, alpha_mu, sigma, X_0, mu, t=None):
    B=np.array([W_t(T, n) for i in range(sigma[mu].size)]) #define array of B (B(t)) in the formula.
    #this should create d=sigma[mu].size (# cols) Brownian motions, each of size n+1
    #We broadcast, NOT matrix multiply (that's what sigma[mu]*B.T does) and sum along d
    return X_0*np.exp((alpha_mu[mu]-0.5*np.sum(np.square(sigma[mu])))*t + np.sum(sigma[mu]*B.T, axis=1))

#FOR V&F SECTION 4.3 ONLY (same as the last function but i plugged the other section 4.3 functions in)
@make_series
def X_mu_43(T, n, sigma, beta, alpha_A, X_0, mu, t=None):
    B=np.array([W_t(T, n) for i in range(sigma[mu].size)]) 
    return X_0*np.exp((calc_alpha_mu_43(mu, sigma, beta, alpha_A)-
                       0.5*np.sum(np.square(sigma[mu])))*t + np.sum(sigma[mu]*B.T, axis=1))

#\hat{\sigma}_mu^a calculation (right side of equation 108)
def calc_sigma_hat(sigma):
    #need to check this
    #sigma.shape[0] is number of rows (N)
    #sum over the rows (this makes the correct shape)
    return sigma-(1/(sigma.shape[0]))*np.sum(sigma, axis=0)

#write these functions tomorrow (discuss with apex pod)
def calc_J(sigma):
    #J dot 1 = 0 (sum of components is 0), J dot sigma^a = 0 for all a (a is column index)
    #>=N-1-d (more if there are linearly dependent columns in sigma)
    return null_space(np.column_stack([sigma, np.ones(sigma.shape[0])]).T)

#FOR V&F SECTION 4.3 ONLY
#equation 110
def calc_alpha_43(sigma, beta, alpha_A):
    return (1/sigma.shape[0])*np.sum(np.matmul(beta, calc_sigma_hat(sigma).T))-np.sum(np.matmul(alpha_A, calc_J(sigma).T))

#FOR V&F SECTION 4.3 ONLY
#equation 108 (left)
def calc_alpha_mu_43(mu, sigma, beta, alpha_A):
    if mu==0:
        return 0
    else:
        return calc_alpha_43(sigma, beta, alpha_A)+np.sum(beta @ calc_sigma_hat(sigma)[mu])+np.sum(alpha_A @ calc_J(sigma)[mu])
```
